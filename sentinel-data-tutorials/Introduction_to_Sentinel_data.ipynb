## Levels of satellite data

Satellite data products are categorized into different levels, each representing a progressively higher level of processing. These levels generally range from 0 to 4, with higher numbers indicating more processed and potentially more user-friendly data. 

| Level | Product Scope  | Geometry  | Time Resolution | Description  | Processing | When to use |
|-------|---------|------|--------|-----|-------|--------|
| L0    | Individual observations   | As recorded by sensor  | Native/orbital        | Raw, unprocessed satellite data, including telemetry and ancillary information                | None; direct from instrument                                            | When you need original, unaltered data for custom processing or calibration |
| L1    | Individual observations   | Geolocated             | Native/orbital        | Radiometrically and geometrically corrected data in sensor units                              | Radiometric correction, calibration, geolocation                       | For sensor-level analysis, calibration checks, or basic radiance studies |
| L2    | Individual observations   | Geolocated             | Native/orbital        | Geophysical variables derived from L1, at original resolution                                  | Algorithmic derivation of physical parameters                          | When you need high-resolution data with exact observation times, for studying local or short-term events. If you plan to apply your own processing or interpolation. |
| L3    | Regional or global product| Regular grid           | Daily, weekly, etc.   | Geophysical variables mapped onto a regular grid to produce regional or global datasets over specific time periods | Spatial and/or temporal aggregation of L1 or L2 data; individual image detail is lost | For relatively high-resolution regional or global studies, when you need data on a standard grid for easier integration and analysis. |
| L4    | Regional or global product| Regular grid           | Multi-day, monthly, etc.| Modelled or assimilated data combining multiple sources and full validation                   | Modelling, data assimilation, interpolation across sensors and time   | When you require a continuous, gap-free dataset, even if it means some loss of the finer detail. |

Each level may be further subdivided into sublevels (e.g., L1A, L1B, L1C), reflecting increasing levels of refinement within than processing level.

> **[satellittdata.no](https://satellittdata.no) hosts a catalogue of Level 0 to Level 2 (L0-L2) Sentinel data.** Level 3 (L3) and Level 4 (L4) data products are provided by downstream data providers.
>

> [Exploring FAIR Data with Python](https://nordatanet.github.io/Exploring_FAIR_data_with_Python)

## Overview of the Sentinel Missions

The **Sentinel satellites** are a series of Earth observation missions developed by the European Space Agency (ESA) as part of the Copernicus Programme. Each mission is designed with specific capabilities to monitor different aspects of our planet.

### Sentinel-1  
Sentinel-1 is a radar imaging mission designed to provide all-weather, day-and-night Earth observation. It primarily measures surface features by sending microwave signals and analysing their reflections, making it ideal for:  
- **Monitoring land surfaces:** detecting changes such as land subsidence, deforestation, and urban development  
- **Ocean observation:** tracking sea ice, oil spills, and wave heights  
- **Emergency response:** rapid imaging after natural disasters like floods and earthquakes  

- **Launches:**  
  - Sentinel-1A: 3 April 2014  
  - Sentinel-1B: 25 April 2016 (mission ended 2022)  
  - Sentinel-1C: 5 December 2024 (restored full mission capabilities, added maritime traffic monitoring)

---

### Sentinel-2  
Sentinel-2 is a multispectral imaging mission capturing high-resolution optical images in visible and near-infrared wavelengths. It focuses on land monitoring by measuring reflected sunlight, enabling:  
- **Vegetation analysis:** tracking crop health, forestry, and plant growth  
- **Soil and water monitoring:** assessing soil moisture and water quality in rivers, lakes, and coastal areas  
- **Disaster management:** mapping areas affected by droughts, wildfires, and floods  

- **Launches:**  
  - Sentinel-2A: 23 June 2015  
  - Sentinel-2B: 7 March 2017  
  - Sentinel-2C: 5 September 2024 (ensures continuity of high-resolution land data)

### Sentinel-3  
Sentinel-3 is a multi-instrument mission that measures various environmental parameters to support oceanography and land monitoring:  
- **Sea-surface topography:** measuring sea level and currents for ocean forecasting  
- **Sea and land surface temperature:** monitoring climate and environmental changes  
- **Ocean and land colour:** assessing water quality, algae blooms, and vegetation health  

- **Launches:**  
  - Sentinel-3A: 16 February 2016  
  - Sentinel-3B: 25 April 2018  

---

### Sentinel-5 Precursor (Sentinel-5P)  
Sentinel-5P monitors atmospheric gases and aerosols that affect air quality and climate, filling the data gap before Sentinel-5’s launch:  
- **Trace gases:** such as nitrogen dioxide (NO₂), ozone (O₃), and carbon monoxide (CO)  
- **Aerosols:** tiny particles affecting air pollution and visibility  

- **Launch:** 13 October 2017 from Plesetsk Cosmodrome, Russia  

---

### Sentinel-4  
Sentinel-4 is an atmospheric monitoring instrument onboard the Meteosat Third Generation Sounder (MTG-S) satellites in geostationary orbit:  
- **Hourly data delivery** on trace gases and pollutants over Europe  
- **Monitors:** ozone, nitrogen dioxide, sulfur dioxide, and aerosols  
- **Purpose:** forecasting and tracking air quality in near real-time  

- **Launch:** 1 July 2025 aboard the first MTG-S satellite  

---

### Sentinel-5  
Sentinel-5 is designed to monitor atmospheric composition from polar orbit aboard MetOp Second Generation satellites:  
- **Measures:** trace gases, aerosols, and air pollutants  
- **Supports:** climate monitoring and air quality forecasting on a global scale  

---

### Sentinel-6  
Sentinel-6 carries a radar altimeter to measure global sea-surface height with high precision:  
- **Primary use:** oceanography and climate studies  
- **Applications:** monitoring sea-level rise, ocean circulation, and climate change impacts  

- **Launch:** 21 November 2020
